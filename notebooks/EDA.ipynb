```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8')

# Load data
employees = pd.read_csv('../data/knowledge_base/structured/employee_directory.csv')
tickets = pd.read_csv('../data/knowledge_base/structured/jira_export.csv')

print(f"Employees: {employees.shape[0]} records")
print(f"Tickets: {tickets.shape[0]} records")
employees.head()
```

```python
# Team Distribution
plt.figure(figsize=(10, 6))
employees['team'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Employee Distribution by Team')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```

```python
# Ticket Priority Analysis
plt.figure(figsize=(12, 5))
sns.countplot(data=tickets, x='priority', hue='status', palette='viridis')
plt.title('Tickets by Priority and Status')
plt.show()

tickets['priority'].value_counts()
```

```python
# Most Active Teams in Tickets
plt.figure(figsize=(10, 6))
tickets['team'].value_counts().head(8).plot(kind='barh', color='coral')
plt.title('Tickets per Team')
plt.xlabel('Number of Tickets')
plt.show()
```

## Key Insights

- Engineering is the largest team
- High volume of High/Critical tickets
- Significant number of blocked tickets across projects
- DevOps team plays critical role in multiple initiatives
